# Software engineering exercise

In this exercise we will create a little software package including tests, starting from simple functions.

1. Create a file (`arithmetic.py`) with two functions: `add` and `multiply` that add and multiply two numbers, respectively, and return the result
2. Now we need to make a package out of this: create a new folder (e.g. `algebra`) and move the file `arithmetic.py` there. To make a package out of it, create the empty file `__init__.py` in this folder.
3. Add the following simple `pyproject.toml` file to the base directory
    ```toml
    [build-system]
    requires = ["setuptools"]
    build-backend = "setuptools.build_meta"

    [project]
    name = "algebra"
    version = "0.1"
    description = "This is an example package."
    authors = [
      {name = "me", email = "me@example.com"}
    ]
    license = {text = "PSF"}
    ```
4. Test the package: run `pip install -e . --user` to install a development version; change to your home directory, start a python shell and try to run `from algebra.arithmetic import add`.
5. Now we want to add tests:
    * create a folder `tests`
    * create a file `test_arithmetic.py` (needs to start with `test_`)
    * add a function for testing the add routine:
```python
    def test_add():
        from algebra.arithmetic import add
        result = add(3, 5)
        assert(result == 8)
```
    * add the following lines to `pyproject.toml`:
```toml
[project.optional-dependencies]
tests = ["pytest"]
```
    * Install the package with automatic dependency resolution `pip install -e .[tests] --user`
    * Run the tests from the base directory with `pytest -v`.
6. Add a test for the multiplication routine

Congratulations! You created a package including tests. Our recommendation is to always create packages for your python software. Moreover, adding tests is very important to validate your code and to ensure that new developments do not break current functionality.
