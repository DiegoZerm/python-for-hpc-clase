## Software Packaging
* Python provides a standard way of software packaging
    * great for "professional" software distribution
    * create zipfiles, tarballs, wheel, or egg packages automatically
    * https://packaging.python.org/
* `setup.py` using the `distutils` or `setuptools` packages
    * Information on version, status, authorship, licensing, dependencies
    * Compilation of Cython, C, Fortran extensions (Makefile replacement)
    * Installation to system, per-user, or arbitrary locations
    * See examples in the folder `setuptools`

### Example: `setup.py` (see folder `setuptools`)

```python
from setuptools import setup
setup(
    name="HelloWorld",
    version="0.1",
    # specify a list of the packages (subdirectory with __init__.py file)
    packages=["HelloWorld"],
    # executable that comes with the package (if applicable)
    scripts=['say_hello.py'],
    # software dependencies (if necessary)
    # install_requires=['docutils>=0.3'],
    # metadata for upload to PyPI
    author="Me",
    author_email="me@example.com",
    description="This is an Example Package",
    license="PSF",
    keywords="hello world example examples",
    url="http://example.com/HelloWorld/",   # project home page, if any
)
```

### Usage of setup.py
* Install package with `python setup.py install --user`<br>
  $\to$ installation path under `~/.local/lib/...`
* This will make the package available for your user
* Build package including extensions with `python setup.py build`
* Create distribution packages for sharing
    * Source package with `python setup.py sdist`
    * Binary package with `python setup.py bdist`

### Advantages of using `setup.py`
* Easy installation to arbitrary locations
* Better handling of different versions
* Easy integration of extensions (Cython, C, Fortran)
* For development: use `pip install -e ./ --user` (development install, always uses the modules from the current folder)
* Facilitates sharing with others

$\to$ recommended for all projects!

### NumPy-style docstrings
* Good for documenting purpose of input variables and result

````python
def add(a, b):
    """
    Add two numbers.
    
    This functions adds two numbers and returns the result.
    
    Parameters
    ----------
    a : int or float
        The first summand.
    b : int or float
        The second summand.
    
    Returns
    -------
    int or float
        The sum of both the input values a and b.
    """
    return a + b
````

### Sphinx Documentation System

* Python documentation generator, de-facto standard
* Create documentation in HTML, PDF, and other formats from
    * reStructuredText (`.rst`) files
    * **Can use docstrings in source files**
* Automated generation of documentation is possible via web services such as _GitHub_ and _Read the Docs_

http://www.sphinx-doc.org/en/master/

### Sphinx example
* Use `sphinx-quickstart` to create default files for the documentation
* Adapt `conf.py` to your needs
    * `sphinx.ext.autodoc`: include docstrings into documentation
    * `sphinx.ext.napoleon`: support for NumPy-style docstrings
* Edit `index.rst`, potentially create other rst files
* Run `make html` to generate the HTML documentation
* Can be integrated into `setup.py` (run ` python setup.py build_sphinx`)
* See the directory `setuptools/hello_world_with_sphinx` for an example

## Unit Tests, Integration Tests
* Tests are a central part of the software development life cycle
    * Unit tests
        * test functions, classes, modules individually
    * Integration tests
        * test the interplay of units, up to the complete application
* `unittest`, `pytest`, and `nose` are popular Python modules for testing
* "continuous integration" allows to run tests automatically at each code commit, e.g. via GitLab-CI runner on a local/virtual machine

### pytest

Here's a minimal example to implement tests for a package "foobar":

1. Create a file `test_foobar.py` with the content

```python
import foobar

def test_foobar_functionality()
    assert(foobar.whatever() == True)
```

2. Run the command `pytest`

Find full documentation at
https://docs.pytest.org/en/latest/contents.html

## Version Control Systems, GIT
* Use version control for _any_ project
    * Distributed development of large projects
    * Small local single-person projects
    * Useful during performance optimization
* `git` is the tool of choice today  
* Good tutorial: https://git-scm.com/docs/gittutorial
* MPCDF GitLab provides web-based repository services for collaboration
    * git repository
    * Issue tracker
    * Wiki

### Continuous Integration

* Tests can be executed when pushing commits
* Tests should be small and fast
* [Usage with gitlab](https://docs.gitlab.com/ce/ci/yaml/README.html): create `.gitlab-ci.yml`
```yaml
test-python:
    script:
      - python setup.py pytest
```
* This will run the tests on every push event
* Uses shared runners to execute the tests $\to$ for special needs, separate runners can be used
* Moreover, documentation can be generated automatically

## PyPI, the Python Package Index
* PyPI is the online software repository for Python
* https://pypi.python.org/pypi
* Contains nearly 140.000 packages (May 2018)
* Anybody may upload packages after registration
* Zip files, tar balls, wheel archives are accepted

### PyPI (cont'd), pip
* `pip` is used to download and install packages
